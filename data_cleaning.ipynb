```python
import pandas as pd
```

```python
df = pd.read_excel('Delinquency_prediction_dataset.xlsx', index_col=0, header=0)
```

```python
df.dtypes
```

```python
df.fillna(0, inplace=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.columns
```

```python
df['Employment_Status'] = df['Employment_Status'].replace({
    'EMP': 'Employed',
    'employed': 'Employed'
})
```

```python
df['Income'].astype(int)
```

```python
df['Credit_Score'].astype(int)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
cat_cols = df.select_dtypes(include=['object']).columns

for col in cat_cols:
    df[col] = df[col].str.strip()
```

```python
df.info()
```

```python
df.shape
```

```python
df['Delinquent_Account'].value_counts()
```

```python
df['Delinquent_Account'].replace({0: 'No', 1: 'Yes'}, inplace=True)
```

```python
df.to_csv('cleaned_dataset.csv', index=False)
```
